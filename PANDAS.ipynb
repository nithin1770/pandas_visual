```python
import pandas as pd
import numpy as np
```

```python
market={
    "vegetables": ["beetroot","carrot","tomato","Kovakkai","coconut"]
     , "color":["purple","orange","red","green","white"], 
     "unit":["1kg","1kg","1kg","1kg","1piece"],
     "min":[15,25,53,30,68],
     "max":[45,50,95,50,43],
    }
```
